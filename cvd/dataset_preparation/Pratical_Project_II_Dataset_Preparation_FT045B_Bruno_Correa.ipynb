```python
# The demographic dataset is the base
merged_df = demographic_df.copy()

# Datasets to merge with demographic one
other_dfs = [
    bpxo_df, bmx_df, alb_cr_df, trigly_df, cbc_df, hdl_df,
    biopro_df, ins_df, hscrp_df, ghb_df, diq_df, alq_df,
    glu_df, tchol_df, smq_df, medical_conditions_df, physical_activity_df
]

# Merging datasets
for df_to_merge in other_dfs:

    # Only select the SEQN (identifier of person) and the columns not present in 'merged_df'
    cols_to_keep = ['SEQN']
    new_cols = [col for col in df_to_merge.columns if col not in merged_df.columns]
    cols_to_keep.extend(new_cols)

    merged_df = merged_df.merge(
        df_to_merge[cols_to_keep],
        on='SEQN',
        how='outer'
    )

display(merged_df.head())
merged_df.info()
```

The fields in Medical Condtions Dataframe related to CVDs are:
* MCQ160b - Ever told had congestive heart failure
* MCQ160c - Ever told you had coronary heart disease
* MCQ160d - Ever told you had angina/angina pectoris
* MCQ160e - Ever told you had heart attack
* MCQ160f - Ever told you had a stroke






```python
# Remove rows with NaN values in the specified columns
merged_df.dropna(subset=['MCQ160B', 'MCQ160C', 'MCQ160D', 'MCQ160E', 'MCQ160F'], inplace=True)

# Create the 'target' column
# A value of 1.0 for these columns indicates a positive response
merged_df['target'] = ((merged_df['MCQ160B'] == 1.0) |
                    (merged_df['MCQ160C'] == 1.0) |
                    (merged_df['MCQ160D'] == 1.0) |
                    (merged_df['MCQ160E'] == 1.0) |
                    (merged_df['MCQ160F'] == 1.0)).astype(int)

# Display the first few rows with the new 'target' column
display(merged_df.info())
display(merged_df.head())

merged_df['target'].value_counts()
```

```python
# Save the merged dataframe to a CSV file
# index=False prevents saving the pandas index (row number column) to the file
merged_df.to_csv('nhanes_2021_2023_merged_data.csv', index=False)
```
